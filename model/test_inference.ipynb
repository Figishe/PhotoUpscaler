```python
import os
from PIL import Image

from file_utils import parse_image_file_paths

from dataset import SuperResDataset
from lit_upscaler import LitSuperResNet
from inference import Inference
```

```python
from dotenv import load_dotenv
load_dotenv()
```

```python
TEST_PATH = os.getenv("TEST_INFERENCE_PATH")

image_file_paths = parse_image_file_paths(TEST_PATH)

model = LitSuperResNet.load_from_checkpoint(
    'checkpoints/v5_d3/upscaler-epoch=096.ckpt',
)

inference = Inference(model)
```

```python
for img_path in image_file_paths:
    img = Image.open(img_path)
    upscaled_img = inference.upscale(img)
    out_path = os.path.join("inference_results", os.path.basename(img_path) + "_upscaled.png")
    upscaled_img.save(out_path, format='PNG')
```
